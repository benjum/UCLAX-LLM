Ollama is a platform for running large language models (LLMs) locally on your machine, offering privacy, performance, and customization.  It supports a wide range of models across different categories, including source models, fine-tuned models, embedding models, and multimodal models.

See:
* Introduction: https://docs.ollama.com/api/introduction
* Quickstart: https://docs.ollama.com/quickstart
* Download: https://ollama.com/download

Basic commands:
* Library of available models: https://ollama.com/library
* List downloaded models: `ollama list`
* Pull a model: `ollama pull <model-name>`
* Run a model: `ollama run <model-name>`
* Turn off thinking in CLI: `/set nothink`

```python
from openai import OpenAI

client = OpenAI(
    base_url='http://localhost:11434/v1/',
    api_key='ollama-local',  # required but ignored
)

chat_completion = client.chat.completions.create(
    messages=[
        {
            'role': 'user',
            'content': 'Say this is a test',
        }
    ],
    model='qwen3:0.6b',
)
print(chat_completion.choices[0].message.content)
```
